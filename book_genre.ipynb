```python
import requests
from collections import defaultdict
import time

```

```python
def get_book_genre(book_title):
    url = "https://www.googleapis.com/books/v1/volumes"
    params = {
        "q": book_title,
        "maxResults": 1,
        "printType": "books",
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = response.json()
        items = data.get("items")
        if items:
            volume_info = items[0].get("volumeInfo", {})
            categories = volume_info.get("categories", ["Bilinmeyen"])
            return categories[0]
    return "Bilinmeyen"
```

```python

book_titles = []
print("Kitap isimlerini girin (bitirmek için boş bir satır bırakın):")
while True:
    title = input("Kitap adı: ").strip()
    if not title:
        break
    book_titles.append(title)
```

```python

genre_dict = defaultdict(list)
```

```python

for title in book_titles:
    print(f"Tür alınıyor: {title}")
    genre = get_book_genre(title)
    genre_dict[genre].append(title)
    time.sleep(1)
```

```python
print("\nKitaplar türlerine göre:")
for genre, titles in genre_dict.items():
    print(f"\n📚 {genre}:")
    for t in titles:
        print(f"  - {t}")
```
